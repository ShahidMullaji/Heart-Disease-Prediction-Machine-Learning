```python
df=pd.get_dummies(df,columns=['Gender', 'Blood Type', 'Medical Condition', 
                              'Insurance Provider', 'Room Number', 'Admission Type', 'Medication', 'Test Results'],drop_first=True
)
```

```python
df=pd.get_dummies(df,columns=['Name', 'Doctor', 'Hospital'],drop_first=True)
df
```

```python
from sklearn.preprocessing import StandardScaler
col=['Age','Billing Amount']
scaler=StandardScaler()
df[col]=scaler.fit_transform(df[col])
df=df.round(2)
```

```python
df['Date of Admission	Billing ']=pd.to_datetime(df['Date of Admission'])
```

```python
df['Discharge Date']=pd.to_datetime(df['Discharge Date'])
```

```python
df['Date of Admission'] = pd.to_datetime(df['Date of Admission'], errors='coerce')
df['Discharge Date'] = pd.to_datetime(df['Discharge Date'], errors='coerce')

print(df[['Date of Admission', 'Discharge Date']].dtypes)   # dono datetime64 aane chahiye

df['length_of_stay'] = (df['Discharge Date'] - df['Date of Admission']).dt.days
```

```python
df['admit_month'] = df['Date of Admission'].dt.month
df['admit_dayofweek'] = df['Date of Admission'].dt.dayofweek 
```
